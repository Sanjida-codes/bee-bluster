```python
import numpy as np

np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])


print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

bonus_marks = marks + 5
print('Bonus Marks:', bonus_marks)


passed_count = np.sum(marks >= 50)
print('Passed Count:', passed_count)

matrix_2x3 = marks.reshape(2, 3)
print('2x3 Matrix:\n', matrix_2x3)
```

```python
import pandas as pd


data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)

print("--- 16. Features & Label ---")
print("Features: study_hours, sleep_hours, attendance, snacks")
print("Label: passed\n")

print("--- 17. Samples and Columns Count ---")
print(f"Total Samples (Rows): {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}\n")

print("--- 18. Students with Attendance >= 80 ---")
high_attendance = df[df['attendance'] >= 80]
print(high_attendance[['student', 'attendance']], "\n")

print("--- 19. Average Study Hours for Students Who Passed ---")
avg_study_passed = df[df['passed'] == 'Yes']['study_hours'].mean()
print(f"Average Study Hours: {avg_study_passed}\n")

print("--- 20. Sorted by Attendance (Descending) ---")
sorted_df = df.sort_values(by='attendance', ascending=False)
print(sorted_df[['student', 'attendance']], "\n")

print("--- 21. Snack Purchases vs Failure Explanation ---")
print("Explanation: No, this tiny dataset does not prove that snack purchases cause failure.")
print("Reason 1: Correlation does not mean causation (eating snacks doesn't directly cause failing).")
print("Reason 2: The sample size (6 students) is far too small to establish any universal rule.")
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)


print('Seeded run 1:   ', generate_scores(True))
print('Seeded run 2:   ', generate_scores(True))
print('Unseeded run 1: ', generate_scores(False))
print('Unseeded run 2: ', generate_scores(False))
```
